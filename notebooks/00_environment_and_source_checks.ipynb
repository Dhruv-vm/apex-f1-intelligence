## 3. Package Availability Audit (Non-Destructive Inspection)

> **Operating Rule:** Inspect presence only. Do NOT execute `pip install` or modify the host environment automatically within this notebook.
> **TODO (Collaborative Step):** Inspect availability of candidate research dependencies.

```python
candidate_packages = [
    "requests",
    "urllib3",
    "pandas",
    "numpy",
    "scipy",
    "fastf1",
    "lightgbm",
    "xgboost",
    "sklearn",
    "pydantic"
]

# TODO: Verify package presence and report versions where available
package_status = {}
for pkg in candidate_packages:
    try:
        mod = __import__(pkg)
        ver = getattr(mod, "__version__", "installed (no __version__)")
        package_status[pkg] = ("AVAILABLE", ver)
    except ImportError:
        package_status[pkg] = ("NOT_INSTALLED", None)

for pkg, (status, ver) in package_status.items():
    version_info = f"({ver})" if ver else ""
    print(f"{pkg:<15}: {status} {version_info}".rstrip())
```

## 4. Robust Repository Path & Directory Accessibility Verification

> **TODO (Collaborative Step):** Discover the project root by searching current and parent directories for project markers (`AGENTS.md`, `README.md`) without relying on `__file__`.

```python
# Robust project-root discovery searching current and parent directories for project markers
current_path = Path.cwd().resolve()
project_root = None

for candidate in [current_path] + list(current_path.parents):
    if (candidate / "AGENTS.md").is_file() and (candidate / "README.md").is_file():
        project_root = candidate
        break

if project_root is None:
    raise FileNotFoundError(f"Could not locate project root containing AGENTS.md and README.md from {current_path}")

print("Discovered Project Root:", project_root)

required_dirs = [
    "docs",
    "data",
    "data/raw",
    "data/processed",
    "scripts",
    "scripts/audit",
    "ml",
    "artifacts",
    "artifacts/reports"
]

for rd in required_dirs:
    target_path = project_root / rd
    exists = target_path.exists()
    status_label = "EXISTS" if exists else "MISSING"
    print(f"{rd:<20}: {status_label} -> {target_path}")
```
